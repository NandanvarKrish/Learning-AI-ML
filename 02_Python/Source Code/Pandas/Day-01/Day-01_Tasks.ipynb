```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)
n = 100

depts  = ['Engineering','Marketing','Finance','HR','Operations']
cities = ['Mumbai','Delhi','Bangalore','Pune','Hyderabad']
levels = ['Junior','Mid','Senior','Lead','Manager']

df = pd.DataFrame({
    'EmpID'   : [f'E{i:03d}' for i in range(1, n+1)],
    'Name'    : [f'Employee_{i}' for i in range(1, n+1)],
    'Age'     : rng.integers(22, 58, n),
    'Dept'    : rng.choice(depts,  n),
    'City'    : rng.choice(cities, n),
    'Level'   : rng.choice(levels, n),
    'Salary'  : rng.integers(35000, 200000, n),
    'YearsExp': rng.integers(1, 20, n),
    'Rating'  : np.round(rng.uniform(2.0, 5.0, n), 1),
    'Remote'  : rng.choice([True, False], n, p=[0.4, 0.6])
})

miss_idx = rng.choice(n, 5, replace=False)
df.loc[miss_idx, 'Rating'] = np.nan

df.to_csv('employees.csv', index=False)
df = pd.read_csv('employees.csv')
print(f"Loaded: {df.shape}")
```

```python

total_emp    = len(df)
num_depts    = df['Dept'].nunique()
num_cities   = df['City'].nunique()

avg_salary   = df['Salary'].mean()
max_salary   = df['Salary'].max()
min_salary   = df['Salary'].min()
med_salary   = df['Salary'].median()
max_emp      = df.loc[df['Salary'].idxmax(), 'Name']
min_emp      = df.loc[df['Salary'].idxmin(), 'Name']

avg_age      = df['Age'].mean()
min_age      = df['Age'].min()
max_age      = df['Age'].max()

remote_count = df['Remote'].sum()
onsite_count = total_emp - remote_count
remote_pct   = remote_count / total_emp * 100
onsite_pct   = onsite_count / total_emp * 100

dept_counts  = df['Dept'].value_counts()

missing_rat  = df['Rating'].isnull().sum()
dup_count    = df.duplicated().sum()


print("""
═══════════════════════════════════════════
       NOVACORP HR ANALYTICS REPORT
═══════════════════════════════════════════

WORKFORCE OVERVIEW""")
print(f"  Total Employees   : {total_emp}")
print(f"  Departments       : {num_depts}")
print(f"  Cities            : {num_cities}")

print(f"""
SALARY ANALYSIS
  Average Salary    : ₹{avg_salary:,.0f}
  Highest Salary    : ₹{max_salary:,}  ({max_emp})
  Lowest Salary     : ₹{min_salary:,}    ({min_emp})
  Median Salary     : ₹{med_salary:,.0f}""")

print(f"""
AGE PROFILE
  Average Age       : {avg_age:.1f} years
  Youngest          : {min_age} years
  Oldest            : {max_age} years""")

print(f"""
REMOTE WORK
  Remote Employees  : {remote_count} ({remote_pct:.1f}%)
  On-site Employees : {onsite_count} ({onsite_pct:.1f}%)""")

print(f"""
DEPARTMENT BREAKDOWN""")
for dept in ['Engineering', 'Marketing', 'Finance', 'HR', 'Operations']:
    count = dept_counts.get(dept, 0)
    print(f"  {dept:<12}: {count} employees")

print(f"""
DATA QUALITY
  Missing Ratings   : {missing_rat} employees
  Duplicates        : {dup_count}

═══════════════════════════════════════════""")
```

```python


salary_lookup = pd.Series(df['Salary'].values, index=df['EmpID'], name='Salary')
print("Salary Lookup Series (first 5):")
print(salary_lookup.head())

def get_salary(emp_id: str) -> float:
    """Return the salary for a given employee ID."""
    if emp_id in salary_lookup.index:
        return salary_lookup[emp_id]
    else:
        return f"❌ EmpID '{emp_id}' not found."

print(f"\nget_salary('E001') → ₹{get_salary('E001'):,}")
print(f"get_salary('E025') → ₹{get_salary('E025'):,}")
print(f"get_salary('E999') → {get_salary('E999')}")

dept_lookup = pd.Series(df['Dept'].values, index=df['EmpID'], name='Dept')

eng_ids = dept_lookup[dept_lookup == 'Engineering'].index.tolist()
print(f"\nEngineering Employee IDs ({len(eng_ids)}):")
print(eng_ids)
```
