```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme(style='whitegrid')
tips = sns.load_dataset('tips')
penguins = sns.load_dataset('penguins')
```

```python
sns.histplot(data=tips, x='total_bill')
```

```python
sns.histplot(data=tips, x='total_bill',
             bins=25,
             kde=True,        
             hue='time',      
             multiple='layer') 
```

```python
sns.histplot(data=tips, x='total_bill', stat='density') 
```

```python
sns.histplot(data=tips, x='total_bill', y='tip', cbar=True)
```

```python
sns.displot(data=tips, x='total_bill', col='time', kde=True, kind='hist')
```

```python
sns.kdeplot(data=tips, x='total_bill')
```

```python
sns.kdeplot(data=tips, x='total_bill', hue='time',
            fill=True,             
            alpha=0.5,
            common_norm=False) 
```

```python
sns.kdeplot(data=tips, x='total_bill', bw_adjust=0.5) 
```

```python
sns.kdeplot(data=tips, x='total_bill', bw_adjust=2)
```

```python
sns.kdeplot(data=penguins, x='bill_length_mm', y='bill_depth_mm',
            fill=True, cmap='rocket')
```

```python
sns.histplot(data=tips, x='total_bill', kde=True, stat='density')
```

```python
sns.ecdfplot(data=tips, x='total_bill')
```

```python
sns.ecdfplot(data=tips, x='total_bill', hue='time')
```

```python
sns.ecdfplot(data=tips, x='total_bill', complementary=True)
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='box')
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='violin')
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='swarm')
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='bar')
```

```python
sns.catplot(data=tips, x='day', y='total_bill',
            hue='sex', col='time',
            kind='box', height=5)
```

```python
sns.catplot(data=tips, x='day', kind='count', hue='time')
```

```python
fig, ax = plt.subplots(figsize=(8,5))
sns.boxplot(data=tips, x='day', y='total_bill', hue='smoker', ax=ax)
```

```python
sns.violinplot(data=tips, x='day', y='total_bill',
               hue='sex', split=True,    
               inner='quartile') 
```

```python
sns.swarmplot(data=tips, x='day', y='total_bill', hue='time', size=4)
```

```python
sns.stripplot(data=tips, x='day', y='total_bill', jitter=True, alpha=0.5)
```

```python
fig, ax = plt.subplots(figsize=(8,5))
sns.boxplot(data=tips, x='day', y='total_bill', color='lightgray', ax=ax)
sns.swarmplot(data=tips, x='day', y='total_bill', color='#6D28D9', size=3, ax=ax)
```

```python
sns.barplot(data=tips, x='day', y='total_bill', hue='sex',
            errorbar='ci',        
            estimator='mean') 
```

```python
sns.countplot(data=tips, x='day', hue='smoker')
```

```python
sns.barplot(data=tips, y='day', x='total_bill', orient='h')
```

```python
sns.barplot(data=tips, x='day', y='total_bill',
            order=['Thur','Fri','Sat','Sun']) 
```
