```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid', palette='Set2')
penguins = sns.load_dataset('penguins').dropna()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(data=penguins, x='body_mass_g', hue='species', multiple='layer', kde=True, ax=axes[0])
axes[0].set_title('Body Mass Distribution (Histplot)')

sns.kdeplot(data=penguins, x='body_mass_g', hue='species', fill=True, common_norm=False, ax=axes[1])
axes[1].set_title('Body Mass Density (KDE)')

sns.ecdfplot(data=penguins, x='body_mass_g', hue='species', ax=axes[2])
axes[2].set_title('Cumulative Distribution (ECDF)')

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))
sns.kdeplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', hue='species', fill=True, levels=5)
plt.title('Bill Dimensions Joint Density (2D KDE)')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=penguins, x='species', y='flipper_length_mm', color='lightgray', ax=ax)
sns.swarmplot(data=penguins, x='species', y='flipper_length_mm', hue='species', palette='Set2', legend=False, ax=ax)
ax.set_title('Flipper Length by Species (Box + Swarm)')
plt.show()

plt.figure(figsize=(8, 5))
sns.violinplot(data=penguins, x='species', y='body_mass_g', hue='sex', split=True, inner='quartile')
plt.title('Body Mass by Species Split by Sex')
plt.show()

sns.catplot(
    data=penguins,
    x='species',
    y='body_mass_g',
    hue='sex',
    col='island',
    kind='bar',
    errorbar='ci'
)
plt.show()
```
