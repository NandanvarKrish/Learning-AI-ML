```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

sns.set_theme(style='whitegrid')
diamonds = sns.load_dataset('diamonds')
diamonds_sample = diamonds.sample(2000, random_state=42)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

sns.regplot(data=diamonds_sample, x='carat', y='price', order=1, ax=axes[0])
axes[0].set_title('Linear Regression (Order 1)')

sns.regplot(data=diamonds_sample, x='carat', y='price', order=2, ax=axes[1])
axes[1].set_title('Polynomial Regression (Order 2)')

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.residplot(data=diamonds_sample, x='carat', y='price', lowess=True, line_kws={'color': 'red'})
plt.title('Residual Plot (Carat vs Price)')
plt.show()

sns.lmplot(data=diamonds_sample, x='carat', y='price', hue='cut', height=5, aspect=1.2)
plt.title('Faceted Regression by Cut')
plt.show()
```

```python
sns.jointplot(data=diamonds_sample, x='carat', y='price', kind='hex')
plt.show()

numeric_cols = diamonds_sample.select_dtypes(include='number')
corr_matrix = numeric_cols.corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    square=True,
    linewidths=0.5
)
plt.title('Numeric Features Correlation Matrix')
plt.show()
```
